```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
load_dotenv()
llm=ChatGoogleGenerativeAI(model="gemma-4-31b-it");
    
```

```python
with open("health_report.txt" , "r") as f:
    blood_report=f.read()

print(blood_report[:200])
```

```python
promt=f"""
    you are data medical data extractor 
    from the blood report below , extract all test value and classify each on as HIGH ,LOW or NORMAL
    base on referrace range providen in report 
    from your response as:
    -Test Name: value | status: HIGH,LOW,NORMAL | Reference: range
    Blood Report:
    {blood_report}
"""
extraction_response=llm.invoke(promt)
extracted_value=extraction_response.text
print("====stage1: extracted value====")
print(extracted_value)
```

```python
diet_prompt=f"""
        you are a clinical   nutisantionst specializing in indian habits.
        Base on the blood work analysis below write.
        1. short hearlth summry in 3 line explining the patient's condtion in simple language 
        2. a short, parctical indian dite plan having two sections (1) food to avid (2) food to eat more of
        do not include any section in dite plan 
        Blood work analiysis:
        {extracted_value}
"""
diet_extraction_response=llm.invoke(diet_prompt)
diet_extraced_value=diet_extraction_response.text
print("=====================stage2: diaet paln and report analysis====================")
print(diet_extraced_value)
      
```
