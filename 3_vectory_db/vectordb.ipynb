```python
import chromadb
client = chromadb.Client()
```

```python
collection = client.create_collection("news_articles")
```

```python
collection.add(
    ids=["id1","id2","id3","id4"],
    documents=[
        "Apple is leading in smart phone game with iphone sale up by 35%",
        "Tesla booked a minor profit of 1 bilion $ in Q2",
        "Apple are high in fiber, vitamin c, and various antioxidants",
        "SpaceX got NASA contract worth 10 bilion $"
    ]
)
```

```python
data = collection.get(include=["documents", "embeddings", "metadatas"])
for i,emb in enumerate(data["embeddings"]):
    print(f"\n Document {i+1}")
    print("Text:", data["documents"][i])
    print("Embading len",len(emb))
    print("first 10 value:",emb[:10])
```

```python
data = collection.query(query_texts=["This is a query related to elon musk"],
                        n_results=2)
print(data)
```

```python
data = collection.query(query_texts=["i love ipad"],n_results=4)
print(data)
```
